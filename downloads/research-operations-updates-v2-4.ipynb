## 物理、数学和信息可用性复核

边界断言和独立公式先于正式结果比对。更完整的golden与扰动测试保存在tests/operations。

```python
# C测试夹具仅验证更新器行为，绝不当作官方数据。
class FixtureResponse:
    status_code=200
    content=b'<html><body>fixture-only</body></html>'
specs=[{'id':'fixture','url':'https://www.jepx.jp/fixture','kind':'html'}]
with tempfile.TemporaryDirectory() as folder:
    target=Path(folder)/'no-writes'
    dry=updater.run_update(specs,dry_run=True,output_directory=target,fetcher=lambda _:FixtureResponse())
    assert not target.exists() and list(Path(folder).iterdir()) == []
    first=updater.run_update(specs,output_directory=folder,fetcher=lambda _:FixtureResponse(),checked_at='2026-10-07T00:00:00Z')
    count=len([p for p in Path(folder).rglob('*') if p.is_file()])
    second=updater.run_update(specs,prior=first,output_directory=folder,fetcher=lambda _:FixtureResponse(),checked_at='2026-10-07T00:00:00Z')
    assert second['changed_sources'] == []
    assert len([p for p in Path(folder).rglob('*') if p.is_file()]) == count
assert not result['scheduling']['new_scheduler_created']
print('Sources:',result['source_count'],'data cutoff:',result['data_freshness'])
print('dry-run no writes and idempotency fixtures passed')
```

```python
# 计算完成之后才读取正式结果作一致性比较；正式结果不作为模型输入。
result['scope'] = result['scope'].replace('188', str(normalized['quality']['study_days']))
if MODULE == 'portfolio':
    result['correlations']['observations'] = normalized['quality']['study_days']
canonical = json.loads(gzip.decompress((root/'research/operations-v2-4/results'/f'{MODULE}.json.gz').read_bytes()))
def semantic_hash(obj):
    return hashlib.sha256(json.dumps(obj,ensure_ascii=False,sort_keys=True,separators=(',',':'),allow_nan=False).encode()).hexdigest()
fresh_sha = semantic_hash(result)
expected_sha = semantic_hash(canonical)
assert fresh_sha == expected_sha, 'fresh result differs from reviewed calculation; investigate before sharing'
(replay/'fresh-result.json.gz').write_bytes(gzip.compress(json.dumps(result,ensure_ascii=False,separators=(',',':'),allow_nan=False).encode(),mtime=0))
print('PASS: raw hashes -> fresh normalization -> fresh model; semantic SHA-256',fresh_sha)
print('No saved model cache was used as input.')
```
